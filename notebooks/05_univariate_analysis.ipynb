# Day 6 — Univariate Analysis

## Objectives

- Analyze individual healthcare variables
- Understand numerical distributions
- Analyze categorical distributions
- Identify central tendency and spread
- Identify possible skewness and unusual observations
- Visualize important variables using Matplotlib and Seaborn
- Document key observations

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style="whitegrid")
```

```python
# load final cleaned dataset
df = pd.read_csv(
    "/content/cleaned_healthcare_patients.csv"
)
df.shape
```

```python
# basic dataset verification
print("Shape:", df.shape)
print("Missing values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())
```

```python
df.head()
```

```python
# age analysis
df['age'].describe()

print("Mean age:", df["age"].mean())
print("Median age:", df["age"].median())
print("Minimum age:", df["age"].min())
print("Maximum age:", df["age"].max())

# visualization
plt.figure(figsize=(8,6))
sns.histplot(
    data=df,
    x="age",
    bins=15,
    kde=True
)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("No of patients")
plt.show()

plt.figure(figsize=(8, 4))

sns.boxplot(
    data=df,
    x="age"
)

plt.title("Age Distribution — Boxplot")
plt.xlabel("Age")

plt.show()
```

```python
# gender analysis
gender_counts = df["gender"].value_counts()
print(gender_counts)

gender_percentage = (
    df["gender"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

print(gender_percentage)

# visualization

plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="gender"
)

plt.title("Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Number of Patients")

plt.show()
```

```python
# blood pressure analysis
df["blood_pressure"].describe()

plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="blood_pressure",
    bins=20,
    kde=True
)

plt.title("Blood Pressure Distribution")
plt.xlabel("Blood Pressure")
plt.ylabel("Number of Patients")

plt.show()

plt.figure(figsize=(8, 4))

sns.boxplot(
    data=df,
    x="blood_pressure"
)

plt.title("Blood Pressure — Boxplot")
plt.xlabel("Blood Pressure")

plt.show()
```

```python
# heart rate analysis
df["heart_rate"].describe()

plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="heart_rate",
    bins=20,
    kde=True
)

plt.title("Heart Rate Distribution")
plt.xlabel("Heart Rate")
plt.ylabel("Number of Patients")

plt.show()

plt.figure(figsize=(8, 4))

sns.boxplot(
    data=df,
    x="heart_rate"
)

plt.title("Heart Rate — Boxplot")
plt.xlabel("Heart Rate")

plt.show()
```
