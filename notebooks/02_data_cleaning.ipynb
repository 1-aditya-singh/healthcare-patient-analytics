## Phase 2 — Day 3: Data Cleaning

### Objective

The objective of Day 3 is to identify and handle missing
values in the healthcare patient dataset.

We will:

- Identify missing values
- Calculate missing percentages
- Understand why missing values matter
- Handle numerical missing values
- Handle categorical missing values
- Verify the cleaned dataset

```python
#  import libraries
import pandas as pd
import numpy as np

print("Libraries imported successfully.")
```

```python
# load dataset
df = pd.read_csv("/content/healthcare_patients.csv")

print("Dataset loaded successfully.")
```

```python
df.shape
```

```python
# check missing values
df.isnull().sum()
```

```python
# calculate missing percentage
missing_percentage = ((df.isnull().sum() / len(df)) *100 ).round(2)
missing_percentage[missing_percentage>0]
```

```python
# Create a missing-value summary
missing_summary = pd.DataFrame({
    'Missing Values': df.isnull().sum(),
    'Missing Percentage': missing_percentage
})
missing_summary = missing_summary[missing_summary['Missing Values'] > 0]
missing_summary
```

```python
# finding the meadian of numearical missing values
numerical_missing_columns = [
    "blood_pressure",
    "heart_rate",
    "bmi",
    "medication_count",
    "treatment_cost"
]

df[numerical_missing_columns].median()
```

```python
# fill missing values with meadian
for column in numerical_missing_columns:
    df[column] = df[column].fillna(df[column].median())

df[numerical_missing_columns].isnull().sum()
```

```python
# Handle categorical missing value
df["smoking_status"].value_counts()
```

```python
df["smoking_status"].value_counts(dropna=False)
```

```python
# Fill categorical missing values
df["smoking_status"].mode()[0]
df["smoking_status"] = df["smoking_status"].fillna(
    df["smoking_status"].mode()[0]
)
```

```python
# verify all missing values
df.isnull().sum()
df.isnull().sum().sum()
```

```python
# verify data set
df.shape
```

```python
# Create a cleaning summary
cleaning_summary = pd.DataFrame({
    "column": [
        "blood_pressure",
        "heart_rate",
        "bmi",
        "medication_count",
        "treatment_cost",
        "smoking_status"
    ],
    "data_type": [
        "Numerical",
        "Numerical",
        "Numerical",
        "Numerical",
        "Numerical",
        "Categorical"
    ],
    "missing_value_strategy": [
        "Median",
        "Median",
        "Median",
        "Median",
        "Median",
        "Mode"
    ]
})

cleaning_summary
```
