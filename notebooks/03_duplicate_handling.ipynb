## Phase 2 — Day 4: Duplicate Record Handling

### Objective

The objective of Day 4 is to identify, inspect, and handle
duplicate patient records.

We will:

- Identify duplicate rows
- Count duplicate records
- Inspect duplicate records
- Remove exact duplicate rows
- Verify the cleaned dataset
- Save the processed dataset

```python
# import libraries
import pandas as pd
import numpy as np

print("Libraries imported successfully.")
```

```python
# load dataset
df = pd.read_csv("/healthcare_patients.csv")

print("Dataset loaded successfully.")
```

```python
df.shape
```

```python
# Check for duplicate rows
is_duplicated = df.duplicated()
is_duplicated.sum()
```

```python
# Inspect the duplicate rows
duplicates = df[df.duplicated()]

duplicates
```

```python
duplicates.shape
```

```python
# Check duplicate patient IDs
duplicated_id = df[df.duplicated()]['patient_id'].sum()

duplicated_id
df["patient_id"].duplicated().sum()
```

```python
duplicate_patient_ids = df[
    df["patient_id"].duplicated(keep=False)
].sort_values("patient_id")

duplicate_patient_ids
```

```python
# Confirm the duplicates are exact
df[df.duplicated(keep=False)].sort_values("patient_id")
```

```python
# remove duplicates
df_cleaned = df.copy()

df_cleaned = df_cleaned.drop_duplicates()

df_cleaned.shape
```

```python
# verify duplicate count
df_cleaned.duplicated().sum()

df_cleaned["patient_id"].duplicated().sum()
```

```python
# Verify the original data wasn't changed
print("Original dataset:", df.shape)
print("Cleaned dataset:", df_cleaned.shape)
```

```python
# Save the processed dataset
import os

os.makedirs("/content/data/processed", exist_ok=True)


output_path = "/content/data/processed/healthcare_patients_no_duplicates.csv"

df_cleaned.to_csv(output_path, index=False)

print("Cleaned dataset saved successfully.")
```

```python
# verify the saved file
pd.read_csv(
    "/content/data/processed/healthcare_patients_no_duplicates.csv"
).shape

pd.read_csv(
    "/content/data/processed/healthcare_patients_no_duplicates.csv"
).duplicated().sum()
```
